# 01 - Data Exploration
## Explore Bajhang Tourism Dataset

```python
import pandas as pd
import os

DATA_DIR = '../data'
files = ['destinations.csv', 'routes.csv', 'culture_festivals.csv', 'hotels_homestays.csv', 'local_guides.csv', 'facilities.csv']
for f in files:
    path = os.path.join(DATA_DIR, f)
    if os.path.exists(path):
        df = pd.read_csv(path)
        print(f'\n=== {f} ===')
        print(f'Shape: {df.shape}')
        print(f'Columns: {list(df.columns)}')
        print(df.head())
```

```python
dest = pd.read_csv(os.path.join(DATA_DIR, 'destinations.csv'))
print('Destination Types:')
print(dest['type'].value_counts())
print('\nRating Stats:')
print(dest['rating'].describe())
```
