# 03 - Data Analysis
## Statistical Analysis of Bajhang Tourism Data

```python
import pandas as pd
import os

DATA_DIR = '../data'
dest = pd.read_csv(os.path.join(DATA_DIR, 'destinations.csv'))
routes = pd.read_csv(os.path.join(DATA_DIR, 'routes.csv'))
culture = pd.read_csv(os.path.join(DATA_DIR, 'culture_festivals.csv'))
hotels = pd.read_csv(os.path.join(DATA_DIR, 'hotels_homestays.csv'))
```

```python
print('=== Altitude Analysis ===')
print(dest['altitude_m'].describe())
print(f'\nAbove 3000m: {len(dest[dest["altitude_m"] > 3000])} destinations')
print(f'Below 1500m: {len(dest[dest["altitude_m"] < 1500])} destinations')
```

```python
print('=== Route Correlation ===')
print(routes[['distance_km', 'duration_days', 'elevation_gain', 'rating']].corr())
```

```python
print('=== Hotel Analysis ===')
print(hotels.groupby('type')['rating'].mean())
print(f'\nTotal rooms: {hotels["rooms"].sum()}')
print(f'Total capacity: {hotels["capacity"].sum()}')
```
