# 04 - Visualization
## Charts and Graphs for Bajhang Tourism Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

DATA_DIR = '../data'
dest = pd.read_csv(os.path.join(DATA_DIR, 'destinations.csv'))
routes = pd.read_csv(os.path.join(DATA_DIR, 'routes.csv'))
culture = pd.read_csv(os.path.join(DATA_DIR, 'culture_festivals.csv'))
hotels = pd.read_csv(os.path.join(DATA_DIR, 'hotels_homestays.csv'))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

dest['type'].value_counts().plot(kind='barh', ax=axes[0,0], color='#2e86c1')
axes[0,0].set_title('Destinations by Type')

dest['altitude_m'].hist(ax=axes[0,1], bins=10, color='#f39c12', edgecolor='white')
axes[0,1].set_title('Altitude Distribution')

routes['difficulty'].value_counts().plot(kind='bar', ax=axes[1,0], color='#27ae60')
axes[1,0].set_title('Routes by Difficulty')

hotels['type'].value_counts().plot(kind='pie', ax=axes[1,1], autopct='%1.1f%%')
axes[1,1].set_title('Hotel Types')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.scatter(routes['distance_km'], routes['elevation_gain'], c=routes['rating'], s=100, cmap='viridis')
plt.colorbar(label='Rating')
plt.xlabel('Distance (km)')
plt.ylabel('Elevation Gain (m)')
plt.title('Route Distance vs Elevation Gain')
plt.show()
```
